# 📊 Progetto  — Analisi dati al Liceo Virgilio

**Autore:** Robert Amariei
---

## Panoramica

Questo progetto raccoglie tre analisi dati condotte per conto della dirigenza del **Liceo Virgilio**, un liceo classico di media dimensione, allo scopo di supportare le decisioni didattiche per il prossimo anno scolastico. Ogni analisi è realizzata come notebook indipendente, con una specifica domanda di partenza, un taglio tecnico proprio e un output pensato come deliverable per il committente.

Il progetto adotta una struttura "stile consulenza": la dirigenza del Liceo è il cliente, e le tre analisi rispondono a richieste concrete di complessità crescente. Lo stack tecnologico è volutamente circoscritto a **NumPy** e **Pandas**, le due librerie fondamentali per l'analisi dati in Python.

## Il contesto: il Liceo Virgilio

Il Liceo Virgilio è un liceo classico situato in una città italiana di medie dimensioni. La scuola conta circa **500 studenti** distribuiti sui cinque anni, con una forte tradizione nelle materie umanistiche (Latino, Greco, Italiano, Storia, Filosofia) e un recente investimento sulle lingue straniere moderne (Inglese, Francese, Spagnolo) e sulle materie scientifiche di base (Matematica, Fisica).

Nel corso dell'ultimo anno accademico, la dirigenza ha maturato la necessità di **sistematizzare l'analisi dei dati didattici** per sostituire le valutazioni aneddotiche con evidenze quantitative. A tale scopo è stato affidato un incarico esterno per condurre **tre analisi distinte**, pensate per rispondere a domande operative diverse:

1. Un'**analisi diagnostica locale** su una singola classe per cogliere criticità specifiche
2. Un'**analisi multidimensionale** sull'intera scuola, per confrontare classi, anni e materie
3. Un **report finale tabellare** da consegnare alla dirigenza come strumento di lavoro ricorrente

## Le tre analisi

### 📘 Notebook 1 — Analisi della classe 3A: una fotografia locale

**Domanda di partenza.** La professoressa di lingue della 3A ha consegnato i voti del primo quadrimestre. La dirigenza chiede un'analisi diagnostica rapida: *"Come sta andando questa classe? Chi è in difficoltà? Quanti studenti stanno veramente rischiando la promozione?"*

**Taglio tecnico.** Analisi su una matrice di piccole dimensioni (5 studenti × 4 materie linguistiche). Uso di NumPy con **loop nidificati**, come richiesto dalla traccia, per produrre analisi per materia, per studente e di insieme.

**Deliverable.** Un mini-report di poche righe con: studenti bocciati per materia, studenti insufficienti in tutte le materie, medie individuali, studenti sopra la sufficienza, tasso di promozione complessivo.

---

### 📗 Notebook 2 — Analisi multidimensionale: voti attraverso classi, materie e anni

**Domanda di partenza.** La dirigenza vuole una **visione d'insieme**, non più limitata a una sola classe. I dati disponibili coprono due anni accademici, tre sezioni del terzo anno e le quattro materie fondamentali del classico. *"Possiamo confrontare le classi? Quali materie vanno meglio dove? C'è una sezione che spicca in positivo o in negativo?"*

**Taglio tecnico.** Lavoro su un **tensore quadridimensionale** (anni × classi × materie × prove), con operazioni di aggregazione lungo assi specifici. Notebook più tecnico dei tre, centrato su operazioni matriciali di NumPy (somma lungo assi, prodotto cartesiano per generazione di scenari, normalizzazione per confronti equi).

**Deliverable.** Matrici aggregate che riassumono il tensore in sintesi leggibili, utili come materiale preparatorio per discussioni in sede di collegio docenti.

---

### 📕 Notebook 3 — Report per la dirigenza: l'analisi diventa tabellare

**Domanda di partenza.** *"Abbiamo bisogno di un formato più leggibile per i nostri incontri settimanali. Possiamo avere tutti questi dati in tabelle strutturate invece che in array NumPy?"* È il momento in cui lo strumento cambia: si passa da NumPy a Pandas, perché i report per utenti non tecnici vivono nel mondo dei DataFrame.

**Taglio tecnico.** Creazione di Series da fonti diverse (lista manuale della segreteria e array automatico del sistema informatico), combinazione in DataFrame, operazioni di intersezione/differenza tra le due fonti di dati, reshape di grandi serie in tabelle, analisi statistiche per colonna.

**Deliverable.** Un DataFrame finale con punteggi per materia, completo di statistiche descrittive (media e deviazione standard per materia): il report effettivamente consegnabile alla dirigenza.

## Nota metodologica

Il progetto combina **task tecnici didattici** (ripresi dalla tracca d'esame di fine corso, come dimostrazione di padronanza di specifici strumenti: loop nidificati, operazioni su assi NumPy, manipolazione di Series e DataFrame) con una **cornice applicativa coerente** (il Liceo Virgilio e le richieste della sua dirigenza).

La scelta di ri-strutturare l'esame come un progetto narrativo anziché come una serie di esercizi isolati risponde a due obiettivi:

1. **Rendere i risultati interpretabili**. Un array di numeri senza contesto è difficile da leggere. Lo stesso array presentato come "i voti della 3A del Liceo Virgilio" diventa immediatamente attribuibile, confrontabile, discutibile.
2. **Mostrare consapevolezza dell'uso pratico degli strumenti**. Ogni task tecnico è motivato dalla domanda di business di turno — con una precisazione importante, descritta qui sotto.

**Trasparenza sui task prevalentemente tecnici.** Alcuni dei task richiesti dalla traccia hanno natura prevalentemente dimostrativa di una tecnica, più che di utilità applicativa diretta. Ad esempio, il **prodotto cartesiano** di vettori casuali (Notebook 2) o l'**individuazione dei multipli di 3** in una serie numerica (Notebook 3) sono esercizi di manipolazione che non rispondono a una domanda di business realistica del Liceo Virgilio. Quando si incontrano questi task, il progetto li presenta onestamente come tali: vengono contestualizzati nel dominio della scuola dove è possibile (es. il prodotto cartesiano per generare scenari di pianificazione), ma viene dichiarato esplicitamente il loro carattere tecnico laddove la forzatura narrativa sarebbe artificiosa.

Questo approccio mira a ottenere il meglio dei due mondi: dimostrare padronanza tecnica sugli strumenti richiesti senza compromettere la credibilità del caso d'uso applicativo.
