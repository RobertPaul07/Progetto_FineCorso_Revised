# 📘 Notebook 1 — Analisi della classe 3A

## Il contesto

La professoressa di lingue della **3A del Liceo Virgilio** ha consegnato alla dirigenza i voti del primo quadrimestre. Si tratta di una classe piccola, cinque studenti, con un quadro didattico che in apparenza preoccupa il consiglio di classe: alcune insufficienze sparse, qualche studente in difficoltà evidente, altri che sembrano tenere il ritmo.

La dirigenza ha richiesto un'analisi diagnostica rapida per avere un quadro quantitativo della situazione, da discutere nel prossimo collegio docenti. Le domande da cui partire sono concrete:

- *Quanti studenti sono in difficoltà in ciascuna materia?*
- *Ci sono studenti in situazione di insufficienza generalizzata?*
- *Qual è la media di ogni studente, e quanti sono al di sopra della sufficienza?*
- *Nel complesso, come sta andando la classe?*

## Approccio tecnico

L'analisi verrà condotta con **NumPy**, utilizzando **loop nidificati** per iterare sugli studenti e sulle materie. È un approccio didatticamente trasparente che rende esplicita la logica di ogni calcolo, e che ben si adatta a un dataset di piccole dimensioni come quello della 3A (5 studenti × 4 materie).

```python
# NOTEBOOK 1 — ANALISI DELLA CLASSE 3A
# Setup: import libreria e caricamento dei dati

import numpy as np

# Elenco degli studenti della 3A del Liceo Virgilio
nomi = np.array(["Robbie", "Marco", "Lore", "Fabri", "Alba"])

# Materie linguistiche del primo quadrimestre
materie = np.array(["Latino", "Spagnolo", "Francese", "Inglese"])

# Matrice dei voti: ogni riga è uno studente, ogni colonna è una materia
# Ordine colonne: Latino, Spagnolo, Francese, Inglese (come nell'array materie)
voti = np.array([
    [8, 8, 9, 4],   # Robbie
    [7, 6, 7, 2],   # Marco
    [10, 7, 8, 9],  # Lore
    [4, 4, 3, 2],   # Fabri
    [9, 8, 9, 6]    # Alba
])

# Verifica rapida delle dimensioni
print(f"Classe: {len(nomi)} studenti")
print(f"Materie monitorate: {len(materie)}")
print(f"Shape della matrice dei voti: {voti.shape}")
```

## Task 1 — Bocciati per materia

Iniziamo con la domanda più immediata: **per ciascuna materia, quanti studenti hanno un voto insufficiente?** Un voto inferiore a 6 è considerato insufficiente, secondo la scala decimale italiana.

L'implementazione itera sulle materie (colonne della matrice `voti`) e per ciascuna conta gli studenti sotto la sufficienza. L'output distingue tra singolare e plurale per una lettura più naturale del report.

```python
# TASK 1 — Bocciati per materia

def mostra_bocciati(nomi, materie, voti):
    """
    Mostra il numero di studenti bocciati per ogni materia.
    Un voto inferiore a 6 è considerato insufficiente.

    Args:
        nomi: array con i nomi degli studenti
        materie: array con i nomi delle materie
        voti: matrice con i voti degli studenti per ogni materia

    Returns:
        None (stampa il risultato a schermo)
    """
    # Converte l'input in un array numpy se non lo è già
    voti_array = np.array(voti)

    # Itera attraverso ogni materia (colonne della matrice)
    for j, materia in enumerate(materie):
        # Conta quanti studenti hanno un voto insufficiente (<6) nella materia corrente
        # voti_array[:, j] seleziona tutti i voti della colonna j (materia corrente)
        bocciati = np.sum(voti_array[:, j] < 6)

        # Stampa il risultato con un messaggio appropriato (singolare o plurale)
        if bocciati == 0:
            print(f"In {materia} nessuno studente ha un voto insufficiente.")
        elif bocciati == 1:
            print(f"In {materia} 1 studente ha un voto insufficiente.")
        else:
            print(f"In {materia} {bocciati} studenti hanno un voto insufficiente.")


# Esecuzione sull'elenco voti della 3A
print("Studenti con voto insufficiente per materia — Classe 3A")
print("-" * 60)
mostra_bocciati(nomi, materie, voti)
```

## Task 2 — Studenti insufficienti in tutte le materie

Il primo task ci ha dato un'immagine "per materia": sappiamo in quali materie si concentra la difficoltà. Ora invertiamo la prospettiva: **ci sono studenti insufficienti in tutte e quattro le materie?** È un'informazione cruciale per la dirigenza, perché identifica gli studenti in **situazione di rischio didattico serio**, che probabilmente necessitano di un piano di recupero strutturato e non solo di interventi puntuali.

Questo task è quello che usa i **loop nidificati** in senso pieno, come richiesto dalla traccia d'esame: il ciclo esterno itera sugli studenti, il ciclo interno sui voti di ciascuno studente. L'uscita anticipata dal ciclo interno tramite `break` è un piccolo accorgimento di efficienza — non appena troviamo un voto sufficiente per uno studente, sappiamo che non è insufficiente ovunque e possiamo fermarci.

```python
# TASK 2 — Studenti insufficienti in tutte le materie

def bocciati_tutto(nomi: np.ndarray, voti: np.ndarray) -> np.ndarray:
    """
    Restituisce gli studenti che hanno tutti i voti inferiori a 6,
    utilizzando loop nidificati.

    Args:
        nomi: array dei nomi degli studenti
        voti: matrice dei voti (righe = studenti, colonne = materie)

    Returns:
        np.ndarray contenente i nomi degli studenti con voti insufficienti
        in tutte le materie
    """
    bocciati = []  # Lista temporanea per accumulare i nomi

    for i in range(len(nomi)):  # Ciclo esterno: sugli studenti
        bocciato = True

        for j in range(len(voti[i])):  # Ciclo interno: sui voti dello studente
            if voti[i][j] >= 6:
                bocciato = False
                break  # Uscita anticipata: basta un voto sufficiente per interrompere

        if bocciato:
            bocciati.append(nomi[i])

    return np.array(bocciati)


# Esecuzione sull'elenco voti della 3A
print("Studenti insufficienti in TUTTE le materie — Classe 3A")
print("-" * 60)

studenti_critici = bocciati_tutto(nomi, voti)

if studenti_critici.size > 0:
    for studente in studenti_critici:
        print(f"- {studente}")
    print(f"\nTotale: {studenti_critici.size} studente/i in situazione di rischio serio")
else:
    print("Nessuno studente ha voti insufficienti in tutte le materie.")
```

## Task 3 — Medie individuali

Passiamo alla dimensione per studente: **qual è la media dei voti di ciascuno?** Questo ribalta ulteriormente la prospettiva: dal "per materia" (Task 1) al "studenti estremi" (Task 2), ora andiamo a una lettura **per studente e individuale**.

Calcoliamo anche la **media della classe**, utile come riferimento per la dirigenza: una classe con media vicina a 6 è in una posizione molto diversa da una con media vicina a 8.

```python
# TASK 3 — Medie individuali e media di classe

def calcolare_media(nomi, voti):
    """
    Calcola e mostra le medie individuali di ogni studente e la media di classe.

    Args:
        nomi: array con i nomi degli studenti
        voti: matrice con i voti (righe=studenti, colonne=materie)

    Returns:
        None (stampa il risultato a schermo)
    """
    # Calcola la media per ogni studente lungo l'asse delle colonne (axis=1)
    medie = np.mean(voti, axis=1)

    # Intestazione del report
    print("Medie individuali — Classe 3A")
    print("-" * 60)

    # Itera su ogni studente e mostra la sua media
    for i, nome in enumerate(nomi):
        media = medie[i]
        # Aggiungiamo un indicatore visivo: 🟢 se sufficiente, 🔴 se insufficiente
        indicatore = "🟢" if media >= 6 else "🔴"
        print(f"{indicatore} {nome}: media {media:.2f}")

    # Calcolo e stampa della media di classe
    media_classe = np.mean(medie)
    print(f"\nMedia della classe: {media_classe:.2f}")


# Esecuzione
calcolare_media(nomi, voti)
```

## Task 4 — Studenti sopra la sufficienza

Come quarto e ultimo task diagnostico: **quanti studenti hanno una media pari o superiore a 6?** È l'indicatore più semplice di "promozione quadrimestrale" basata sulla media, sebbene nella pratica la promozione finale dipenda da molti altri fattori (comportamento, specifiche soglie per materia, giudizio del consiglio di classe).

Useremo il risultato già calcolato nel Task 3 (l'array delle medie individuali), evitando ricalcoli inutili.

```python
# TASK 4 — Studenti con media sopra la sufficienza

# Riutilizziamo il calcolo delle medie già fatto nel Task 3
medie = np.mean(voti, axis=1)

# Maschera booleana: True se la media è >= 6
sopra_sufficienza = medie >= 6

# Intestazione report
print("Studenti con media sopra la sufficienza — Classe 3A")
print("-" * 60)

# Itera e mostra solo quelli che soddisfano la condizione
for i, nome in enumerate(nomi):
    if sopra_sufficienza[i]:
        print(f"- {nome}: media {medie[i]:.2f}")

# Statistica finale di sintesi
num_sopra = np.sum(sopra_sufficienza)
totale = len(nomi)
percentuale = (num_sopra / totale) * 100
print(f"\nTotale: {num_sopra}/{totale} studenti ({percentuale:.1f}%) hanno media sopra la sufficienza")
```

## 🎓 Extra — Tasso di promozione della classe

I quattro task precedenti hanno risposto alle domande diagnostiche della dirigenza. Aggiungiamo ora un'**analisi extra** che offre una visione di sintesi finale: **il tasso di promozione della classe**, calcolato secondo il criterio più rigoroso possibile — *uno studente è considerato "promosso" solo se ha voto sufficiente in TUTTE le materie*.

Si tratta di un criterio volutamente severo (nella pratica, la promozione considera il profitto complessivo e non richiede la sufficienza in ogni singola prova) ma utile come **indicatore di allarme**: se il tasso di promozione calcolato in questo modo è basso, significa che una parte significativa della classe ha almeno una materia problematica, anche quando complessivamente la media è buona.

Questo extra rappresenta il valore aggiunto dell'analisi rispetto alla traccia iniziale: trasforma la diagnosi in una misura sintetica che la dirigenza può usare come KPI di classe.

```python
# EXTRA — Tasso di promozione della classe

def calcolare_promossi(nomi, voti):
    """
    Determina quali studenti sono promossi in base ai loro voti e visualizza
    un report di sintesi con il tasso di promozione della classe.

    Uno studente è considerato promosso se ha un voto sufficiente (>=6)
    in TUTTE le materie monitorate.

    Args:
        nomi: array con i nomi degli studenti
        voti: matrice con i voti (righe=studenti, colonne=materie)

    Returns:
        None (stampa il report a schermo)
    """
    # Array booleano: True se lo studente ha TUTTI i voti sufficienti
    # np.all(condizione, axis=1) verifica che la condizione sia vera per
    # tutti gli elementi di ogni riga
    promossi = np.all(voti >= 6, axis=1)

    # Conteggi
    num_promossi = np.sum(promossi)
    num_studenti = len(nomi)
    num_non_promossi = num_studenti - num_promossi

    # Intestazione del report
    print(f"Tasso di promozione — Classe 3A ({num_promossi}/{num_studenti} promossi)")
    print("-" * 60)

    # Elenco studenti promossi
    print("Studenti con sufficienza in tutte le materie:")
    for i, p in enumerate(promossi):
        if p:
            print(f"  🟢 {nomi[i]}")

    # Elenco studenti non promossi (solo se ce ne sono)
    if num_non_promossi > 0:
        print("\nStudenti con almeno un'insufficienza:")
        for i, p in enumerate(promossi):
            if not p:
                print(f"  🔴 {nomi[i]}")

    # Indicatore sintetico finale
    percentuale = (num_promossi / num_studenti) * 100
    print(f"\nTasso di promozione della 3A: {percentuale:.1f}%")


# Esecuzione
calcolare_promossi(nomi, voti)
```

## 📋 Sintesi per la dirigenza

L'analisi diagnostica della classe 3A fornisce il seguente quadro, riassunto in quattro punti operativi:

**1. La difficoltà è concentrata in Inglese.** Tre studenti su cinque hanno voti insufficienti in Inglese, contro un solo studente in Latino, Spagnolo e Francese. È il segnale di un **problema di materia**, non di classe: la dirigenza dovrebbe valutare un intervento mirato sull'insegnamento dell'Inglese (ore di recupero dedicate, supporto al docente, o verifica della coerenza con i prerequisiti degli studenti in ingresso).

**2. Un solo studente è in difficoltà generalizzata.** Fabri è insufficiente in tutte e quattro le materie linguistiche. Questo è lo studente che richiede **attenzione prioritaria** con un piano di supporto personalizzato, possibilmente in coordinamento con la famiglia.

**3. La classe nel suo complesso tiene.** Quattro studenti su cinque hanno una media ≥ 6, con tre di loro (Robbie, Lore, Alba) ben al di sopra del 7. Marco è al limite della sufficienza e va monitorato, ma non è in situazione di allarme.

**4. Il tasso di promozione severo è del 40%.** Tra i quattro studenti con media sufficiente, **solo due hanno sufficienza in tutte le materie**. È un indicatore che, applicato su scala scolastica, potrebbe essere utile come KPI per identificare classi con criticità diffuse di profitto.
